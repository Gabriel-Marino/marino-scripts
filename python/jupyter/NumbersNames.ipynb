```python
unidades = ["zero", "um", "dois", "três", "quatro", "cinco", "seis", "sete", "oito", "nove", "dez", "onze", "doze", "treze", "quatorze", "quinze", "dezesseis", "dezessete", "dezoito", "dezenove"]
dezenas = ["dez", "vinte", "trinta", "quarenta", "cinquenta", "sessenta", "setenta", "oitenta", "noventa"]
```

```python
# for i, j in list(map(lambda x, y: (x, y), list(range(10)), list(range(10)))):
for i in range(10):
    for j in range(10):
        if i*10+j < 20:
            print(f"{i*10+j} = {unidades[i*10+j]}")
        elif j == 0:
            print(f"{i*10+j} = {dezenas[i-1]}")
        else:
            print(f"{i*10+j} = {dezenas[i-1]} e {unidades[j]}")
```

```python
NumberName = dict()
for i in range(10):
    for j in range(10):
        if i*10+j < 20:
#            print(f"{i*10+j} = {unidades[i*10+j]}")
            NumberName.update({f"{i*10+j}": f"{unidades[i*10+j]}"})
        else:
#            print(f"{i*10+j} = {dezenas[i-1]}" + (f" e {unidades[j]}" if j > 0 else ""))
            NumberName.update({f"{i*10+j}": f"{dezenas[i-1]}" + (f" e {unidades[j]}" if j > 0 else "")})

NumberNamesAlph = dict(sorted(NumberName.items(), key=lambda x: x[1]))
data = [(int(key), list(NumberNamesAlph.keys()).index(key)) for key in NumberNamesAlph.keys()]
```

```python
import matplotlib.pyplot as plt 

x, y = zip(*data)
plt.figure(figsize=(11,7))
plt.scatter(x, y, marker='o', s=100)
plt.xlabel("Number name")
plt.ylabel("Position when alphabetically ordered")
plt.title("Distribution of the alphabetically ordered numbers in Portuguese")

for x, y in data:
    if x < 20 or x%10==0:
        plt.annotate(x, (x, y), textcoords="offset points", xytext=(0, 0), ha='center', va='center', fontsize=7, color='white', fontweight='bold')

plt.show()
```
