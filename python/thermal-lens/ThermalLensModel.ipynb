$$
\frac{\partial\,Q}{\partial\,t}=-\oint_{\partial\Omega}\vec{q}\cdot\hat{n}\,dS.
$$

$$
dU=\delta\,Q-p\,dV.
$$

$$
\begin{aligned}
dU&=\left(\frac{\partial\,U}{\partial\,T}\right)_{p}dT+\left(\frac{\partial\,U}{\partial\,p}\right)_{T}dp, \\
% \delta\,Q&=\left(\frac{\partial\,Q}{\partial\,T}\right)_{p}dT+\left(\frac{\partial\,Q}{\partial\,p}\right)_{T}dp,  \\
dV&=\left(\frac{\partial\,V}{\partial\,T}\right)_{p}dT+\left(\frac{\partial\,V}{\partial\,p}\right)_{T}dp.
\end{aligned}
$$

$$
\partial\,Q\approx\delta\,Q=dU+p\,dV.
$$

\begin{equation}
\begin{aligned}
\partial\,Q&\approx\left(\frac{\partial\,U}{\partial\,T}\right)_{p}dT+\left(\frac{\partial\,U}{\partial\,p}\right)_{T}dp+p\left(\left(\frac{\partial\,V}{\partial\,T}\right)_{p}dT+\left(\frac{\partial\,V}{\partial\,p}\right)_{T}dp\right) \\
&=\left(\frac{\partial\,U}{\partial\,T}+p\frac{\partial\,V}{\partial\,T}\right)_{p}dT+\left(\frac{\partial\,U}{\partial\,p}+p\frac{\partial\,V}{\partial\,p}\right)_{T}dp \\
&=\left(\frac{\partial\,U}{\partial\,T}+p\frac{\partial\,V}{\partial\,T}\right)_{p}dT=C_p\,dT\ (\text{Adiabatic process, thus }dp=0).
\end{aligned}
\end{equation}

$$
\oint_{\partial\Omega}\vec{q}\cdot\hat{n}\,dS=\iiint_\Omega\vec{\nabla}\cdot\vec{q}\,dV.
$$

$$
C_p\frac{\partial\,T}{\partial\,t}=-\iiint_\Omega\vec{\nabla}\cdot\vec{q}\,dV.
$$

$$
C_p=c_p\,m=c_p\iiint_\Omega\,\rho\,dV=\iiint_\Omega\,c_p\,\rho\,dV
$$

$$
\iiint_\Omega\,c_p\,\rho\frac{\partial\,T}{\partial\,t}\,dV=-\iiint_\Omega\vec{\nabla}\cdot\vec{q}\,dV.
$$

$$
c_p\,\rho\frac{\partial\,T}{\partial\,t}=-\vec{\nabla}\cdot\vec{q}.
$$

$$
\vec{q}=\vec{q}_{advective}+\vec{q}_{conductive}+\vec{q}_{radiative}.
$$

$$
\begin{aligned}
\vec{q}_{conductive}&=-\kappa\,\vec{\nabla}T,  \\
\vec{q}_{advective}&=0\ (\text{choosen particullarly}),  \\
\vec{q}_{radiative}&=\vec{q}_{ext}.
\end{aligned}
$$

$$
c_p\,\rho\frac{\partial\,T}{\partial\,t}=\vec{\nabla}\cdot(\kappa\,\vec{\nabla}T-\vec{q}_{ext}).
$$

$$
c_p\,\rho\frac{\partial\,T}{\partial\,t}=\kappa\,\nabla^{2}T-S\ (S=\vec{\nabla}\cdot\vec{q}_{ext}\text{, named the source}).
$$

Here we choose as a source the heat delivered by a continuos $TEM_{00}$ Gaussian beam, that is,
$$
S(r,z)=I(r)\cdot\exp\{-\beta\,z\}=\frac{2\,P_e\,\phi}{\pi\,\omega_{e0}^2}\exp\left\{-\frac{2\,r^2}{\omega_{e0}^2}\right\}\beta\exp\left\{-\beta\,z\right\}.
$$

$$
\lim_{\beta\rightarrow0}\exp\{-\beta\,z\}=1\therefore S(r)=\frac{2\,P_e\,\phi\,\beta}{\pi\,\omega_{e0}^2}\exp\left\{-\frac{2\,r^2}{\omega_{e0}^2}\right\}\ (\beta<<1).
$$

$$
c_p\,\rho\frac{\partial\,T}{\partial\,t}=\kappa\,\nabla^{2}T-\frac{2\,P_e\,\phi\,\beta}{\pi\,\omega_{e0}^2}\exp\left\{-\frac{2\,r^2}{\omega_{e0}^2}\right\}.
$$

The boundary and initial conditions are taken as
$$
\begin{aligned}
T[r,z,t=0]=0,  \\
\lim_{r\rightarrow\infty}T[r,z,t]=0,  \\
\left.\frac{\partial\,T[r,z,t]}{\partial\,z}\right|_{z=0}=0.
\end{aligned}
$$

As the heat conduction is being solved for a continuos $TEM_{00}$ Gaussian beam, it is taken that the thermal wave T is azimuthally symmetrical in a cylindrical coordinate system following the symmetry of the beam, thus

$$
\frac{c_p\,\rho}{\kappa}\frac{\partial\,T}{\partial\,t}=\frac{1}{r}\frac{\partial}{\partial\,r}\left(r\frac{\partial\,T}{\partial,r}\right)+\frac{\partial^2\,T}{\partial\,z^2}-\frac{2\,P_e\,\phi\,\beta}{\pi\,\kappa\,\omega_{e0}^2}\exp\left\{-\frac{2\,r^2}{\omega_{e0}^2}\right\}.
$$

Taking advantage of this symmetry and of the boundary and initial conditions, it is chosen to use integral transforms to solve the equation. Being chosen, the Hankel transform $H_n\{f(r)\}(\alpha)$ for the radial coordinate, the Fourier-cosine $F_c\{f(z)\}(k)$ transform for the longitudinal coordinate and the Laplace $L\{f(t)\}(s)$ transform for the time coordinate. Specially because of their derivative properties,
$$
\begin{aligned}
H_n\left\{\frac{1}{r}\frac{\partial}{\partial\,r}\left(r\frac{\partial\,f}{\partial,r}\right)-\frac{n^2}{r^2}f\right\}&=-\alpha^2\,H_n\{f(r)\}(\alpha),  \\
F_c\left\{\frac{\partial^2\,f}{\partial\,z^2}\right\}&=-k^2\,F_c\{f(z)\}(k)-\sqrt{\frac{2}{\pi}}\left.\frac{\partial\,f}{\partial\,z}\right|_{z=0},  \\
L\left\{\frac{\partial\,f(t)}{\partial\,t}\right\}&=s\,L\{f(t)\}(s)-f(t=0).
\end{aligned}
$$
Particulartly, we have
$$
\begin{aligned}
H_0\left\{\frac{1}{r}\frac{\partial}{\partial\,r}\left(r\frac{\partial\,T[r,z,t]}{\partial,r}\right)-\frac{0}{r^2}f\right\}&=-\alpha^2\,H_0\{T[r,z,t]\}(\alpha)=-\alpha^2\,T_0[\alpha,z,t],  \\
F_c\left\{\frac{\partial^2\,T[r,z,t]}{\partial\,z^2}\right\}&=-k^2\,F_c\{T[r,z,t]\}(k)-\sqrt{\frac{2}{\pi}}\left.\frac{\partial\,T[r,z,t]}{\partial\,z}\right|_{z=0}=-k^2\,T[r,k,t],  \\
L\left\{\frac{\partial\,T[r,z,t]}{\partial\,t}\right\}&=s\,L\{f(t)\}(s)-f(t=0)=s\,T[r,z,s].
\end{aligned}
$$

$$

$$
L\left\{F_c\left\{H_0\left\{\frac{1}{D_T}\frac{\partial\,T[r,z,t]}{\partial\,t}\right\}\right\}\right\}=L\left\{F_c\left\{H_0\left\{\frac{1}{r}\frac{\partial}{\partial\,r}\left(r\frac{\partial\,T[r,z,t]}{\partial,r}\right)+\frac{\partial^2\,T[r,z,t]}{\partial\,z^2}-\frac{2\,P_e\,\phi\,\beta}{\pi\,\kappa\,\omega_{e0}^2}\exp\left\{-\frac{2\,r^2}{\omega_{e0}^2}\right\}\right\}\right\}\right\}.
$$

$$
\frac{1}{D_T}L\left\{\frac{\partial\,T_0[\alpha,k,t]}{\partial\,t}\right\}=H_0\left\{\frac{1}{r}\frac{\partial}{\partial\,r}\left(r\frac{\partial\,T[r,k,s]}{\partial,r}\right)\right\}+F_c\left\{\frac{\partial^2\,T_0[\alpha,z,s]}{\partial\,z^2}\right\}-\frac{2\,P_e\,\phi\,\beta}{\pi\,\kappa\,\omega_{e0}^2}H_0\left\{\exp\left\{-\frac{2\,r^2}{\omega_{e0}^2}\right\}\right\}F_c\left\{1\right\}L\left\{1\right\}.
$$

$$
\frac{s}{D_T}T_0[\alpha,k,s]=-\alpha^2\,T_0[\alpha,k,s]-k^2\,T_0[\alpha,k,s]-\frac{2\,P_e\,\phi\,\beta}{\pi\,\kappa\,\omega_{e0}^2}H_0\left\{\\exp\left\{-\frac{2\,r^2}{\omega_{e0}^2}\right\}\right\}DiracDelta[k]\frac{1}{s}.
$$

$$
\left(\frac{s}{D_T}+\alpha^2+k^2\right)\,T_0[\alpha,k,s]=-\frac{2\,P_e\,\phi\,\beta}{\pi\,\kappa\,\omega_{e0}^2}H_0\left\{\exp\left\{-\frac{2\,r^2}{\omega_{e0}^2}\right\}\right\}DiracDelta[k]\frac{1}{s}.
$$

$$
T_0[\alpha,k,s]=-\frac{2\,P_e\,\phi\,\beta}{\pi\,\kappa\,\omega_{e0}^2}\frac{DiracDelta[k]}{s\left(\frac{s}{D_T}+\alpha^2+k^2\right)}H_0\left\{\exp\left\{-\frac{2\,r^2}{\omega_{e0}^2}\right\}\right\}.
$$

$$
T_0[\alpha, k=0,s]=-\frac{2\,P_e\,\phi\,\beta\,D_T}{\pi\,\kappa}\frac{1}{\left(s^2+D_T\,s\,\alpha^2\right)}H_0\left\{\frac{1}{\omega_{e0}^2}\exp\left\{-\frac{2\,r^2}{\omega_{e0}^2}\right\}\right\}.
$$

```python
r, α = sm.symbols("r α", real=True, positive=True)
ν = sm.symbols("ν", integer=True, positive=True)
def HankelTransform(f, ν: int = 0):
    return sm.integrate(f * sm.besselj(ν, α * r) * r, (r, 0, sm.oo))
```

```python
import numpy as np
from scipy.special import expi
from scipy.integrate import quad_vec
import matplotlib.pyplot as plt

mDef = lambda rp1, re0: rp1**2/re0**2
VDef = lambda z1, z2, zcp: z1/zcp + (zcp/z2) * (1+(z1/zcp)**2)
rDef = lambda z, z0, zc, r0: r0 * np.sqrt(1+((z-z0)/zc)**2)
zcDef = lambda r0, wl: np.pi*(r0**2)/wl
tcDef = lambda re0, DT: re0**2/(4*DT)
DTDef = lambda κ, ρ, cp: κ/(ρ*cp)

m = lambda z1, wlp, rp0, re0: mDef(rDef(z1, 0, zcDef(rp0, wlp), rp0), re0)
V = lambda z1, z2, wlp, rp0: VDef(z1, z2, zcDef(rp0, wlp))
tc = lambda re0, κ, ρ, cp: tcDef(re0, DTDef(κ, ρ, cp))

phase = lambda g, t, tc, m, θ: (θ/2)*(expi(-2*g*m)-expi(-2*g*m*(1+2*t/tc)))
field = lambda g, t, tc, m, V, θ: np.exp(-(1+1j*V)*g)*np.exp(-1j*phase(g, t, tc, m, θ))

def LightIntensity(t, z1, z2, wlp, rp0, re0, κ, ρ, cp, θ):
    tc_val = tc(re0, κ, ρ, cp)
    m_val = m(z1, wlp, rp0, re0)
    V_val = V(z1, z2, wlp, rp0)
    int_factor = lambda g: field(g, t, tc_val, m_val, V_val, θ)
    res, _ = quad_vec(int_factor, 0, np.inf, limit=200)
    return (abs(res)**2)/(1+V_val**2)

t_val = np.linspace(0.01, 1.0, 1000)
I_val = [LightIntensity(t, z1=1.E-1, z2=3., wlp=632.E-9, rp0=69.E-6, re0=24.E-6, κ=.598, ρ=997048, cp=4.18, θ=.09) for t in t_val]

m_val = m(z1=1.E-1, wlp=632.E-9, rp0=69.E-6, re0=24.E-6)
V_val = V(z1=1.E-1, z2=3., wlp=632.E-9, rp0=69.E-6)

plt.plot(t_val, I_val)
plt.xlabel("Time (seconds)")
plt.ylabel("Light intensity (A. U.)")
plt.title("Thermal lens light intensity")
plt.text(
    0.9, 0.4, f"m = {m_val:.2f}\nV = {V_val:.2f}",
    transform=plt.gca().transAxes,
    fontsize=10,
    verticalalignment='center',
    horizontalalignment='left',
    bbox=dict(boxstyle='round,pad=0.5', facecolor='lightgray', alpha=0.7)
)
plt.show()
```
